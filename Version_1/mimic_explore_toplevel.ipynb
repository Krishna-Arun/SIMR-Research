# MIMIC-IV Demo — Data Exploration
100-patient demo of Beth Israel Deaconess Medical Center EHR data.

```python
import pandas as pd

DATA = "mimic-iv-clinical-database-demo-2.2"
HOSP = f"{DATA}/hosp"
ICU  = f"{DATA}/icu"
```

## Hospital module

```python
admissions = pd.read_csv(f"{HOSP}/admissions.csv.gz")
print(admissions.shape)
admissions.head()
```

```python
patients = pd.read_csv(f"{HOSP}/patients.csv.gz")
print(patients.shape)
patients.head()
```

```python
diagnoses = pd.read_csv(f"{HOSP}/diagnoses_icd.csv.gz")
d_icd = pd.read_csv(f"{HOSP}/d_icd_diagnoses.csv.gz")
# Join codes with their human-readable descriptions
diagnoses_named = diagnoses.merge(d_icd, on=["icd_code", "icd_version"], how="left")
print(diagnoses_named.shape)
diagnoses_named.head(10)
```

```python
labevents = pd.read_csv(f"{HOSP}/labevents.csv.gz")
d_lab = pd.read_csv(f"{HOSP}/d_labitems.csv.gz")
print(labevents.shape)
labevents.head()
```

## ICU module

```python
icustays = pd.read_csv(f"{ICU}/icustays.csv.gz")
print(icustays.shape)
icustays.head()
```

```python
# chartevents is the largest table — vitals, GCS, etc.
chartevents = pd.read_csv(f"{ICU}/chartevents.csv.gz")
d_items = pd.read_csv(f"{ICU}/d_items.csv.gz")
print(chartevents.shape)
chartevents.head()
```

## Linking hosp + ICU for a single patient

```python
# Pick the first patient and pull their full picture
sample_id = patients["subject_id"].iloc[0]
print(f"Patient: {sample_id}")

print("\n--- Admissions ---")
display(admissions[admissions.subject_id == sample_id])

print("\n--- ICU stays ---")
display(icustays[icustays.subject_id == sample_id])

print("\n--- Diagnoses ---")
display(diagnoses_named[diagnoses_named.subject_id == sample_id][["seq_num", "icd_code", "long_title"]])
```
